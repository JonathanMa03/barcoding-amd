We want to test e2e files since they are preferable to JPGs by preserving more data

```python
import eyepy as ep
import matplotlib.pyplot as plt

e2e_path = "../data/heyex/096/008001.E2E"

ev = ep.import_heyex_e2e(e2e_path)

print(type(ev))
print(ev.data.shape)
print(ev.layers.keys())
```

```python
bscan_idx = 50

plt.figure(figsize=(10, 5))
plt.imshow(ev.data[bscan_idx], cmap="gray")
plt.title(f"B-scan {bscan_idx}")
plt.axis("off")
plt.show()
```

```python
print(ev.meta.keys() if isinstance(ev.meta, dict) else ev.meta)
print(dir(ev))
print(ev.slabs)
print(ev.volume_maps)
```

```python
meta_text = str(ev.meta).lower()

for term in ["overlay", "measure", "distance", "annotation", "mark", "line"]:
    print(term, term in meta_text)
```
